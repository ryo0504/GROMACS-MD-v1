# Molecule exploration
教育用ethanolのexplicit-H構造。seedを指定して生成します。MACEは確認したmodelでのみ使います。

````python
from pathlib import Path
import uuid
from mdworkbench.chemistry.build import build_molecule, write_sdf
from mdworkbench.chemistry.convert import to_ase
mol = build_molecule("CCO", seed=20261004)
atoms = to_ase(mol)
atoms
````

````python
from rdkit.Chem import Draw
Draw.MolToImage(mol)
````

````python
from ase.io import write
root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
output = root / "runs" / ("notebook-build-" + uuid.uuid4().hex[:8])
output.mkdir(parents=True)
write_sdf(mol, output / "molecule.sdf")
write(output / "molecule.xyz", atoms)
output
````
