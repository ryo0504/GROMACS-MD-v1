# Trajectory analysis
入力path/selectionを対象runに合わせます。PBC補正とreferenceを確認してください。MDAnalysisの距離はÅです。

```python
from pathlib import Path
from mdworkbench.analysis.trajectory import load_trajectory, rmsd, read_xvg
from mdworkbench.analysis.rdf import rdf
root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
run = root / "runs/example-001"
topology = run / "production/production.tpr"
trajectory = run / "production/production.xtc"
if topology.exists() and trajectory.exists():
    universe = load_trajectory(topology, trajectory)
    print(universe)
else:
    universe = None
    print("Choose a completed run before analysis.")
```

```python
import matplotlib.pyplot as plt
if universe is not None:
    # Make whole/unwrap first if needed for the selected system.
    values = rmsd(universe, selection="all")
    plt.plot(values[:, 1], values[:, 2])
    plt.xlabel("Time (ps)")
    plt.ylabel("RMSD (Å)")
```

```python
if universe is not None:
    # Example selections; choose atom names and exclusion policy for your topology.
    names = {a.name for a in universe.atoms}
    if "O1" in names and "C1" in names:
        bins, values = rdf(universe, "name O1", "name C1", distance_range=(0, 10))
        plt.plot(bins, values)
        plt.xlabel("r (Å)")
        plt.ylabel("g(r)")
```

```python
energy = run / "production/energy.xvg"
# Export chosen terms using gmx energy first; selection is explicit and interactive.
if energy.exists():
    data = read_xvg(energy)
    plt.plot(data[:, 0], data[:, 1])
    plt.xlabel("Time (ps)")
    plt.ylabel("Chosen energy term (see XVG header)")
```
